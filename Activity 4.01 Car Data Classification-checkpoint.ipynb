```python
# Load the dataset from the web into a pandas DataFrame
import pandas as pd
```

```python
file_url = 'https://raw.githubusercontent.com/PacktWorkshops/The-Applied-Artificial-Intelligence-Workshop/master/Datasets/car.csv'
```

```python

df = pd.read_csv(file_url)
```

```python
# Show the first few rows of the data
df.head()
```

```python
from sklearn import preprocessing
```

```python
# Function to convert text values into numbers so the model can use them
def encode(data_frame, column):
    label_encoder = preprocessing.LabelEncoder()
    label_encoder.fit(data_frame[column].unique())
    return label_encoder.transform(data_frame[column])
```

```python
# Apply the encoding to every column in the dataset
for column in df.columns:
    df[column] = encode(df, column)
```

```python
# Display the encoded dataset
df.head()
```

```python
# Separate the target column (class) from the input features
label = df.pop('class')
```

```python

from sklearn import model_selection
```

```python
features_train, features_test, label_train, label_test = model_selection.train_test_split(df, label, test_size=0.1, random_state=88)
# Split data into training and testing sets (90% training, 10% testing)
```

```python

from sklearn.tree import DecisionTreeClassifier
```

```python

# Create a decision tree model
decision_tree = DecisionTreeClassifier()
```

```python
# Train the model using the training data
decision_tree.fit(features_train, label_train)
```

```python
# Check how well the model performs on the test data
decision_tree.score( features_test, label_test )
```

```python
from sklearn.metrics import classification_report
```

```python
print(classification_report(label_test, decision_tree.predict(features_test)))
# Print precision, recall, and F1-score for each class
```

```python
#About 95% of the time, the decision tree model was able to accurately figure out the different categories of car acceptability.
#From the classification report, it is evident that most classes have a high precision and recall, 
# hence the model is a dependable tool for the car rental company in picking out cars that will satisfy customers.
```
